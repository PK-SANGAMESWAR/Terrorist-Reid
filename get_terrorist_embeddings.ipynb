```python
from get_person_crop_embedding import get_osnet_1x_embedding
import os
import torch
import numpy as np
import pandas as pd
```

```python
def get_avg_embedding(person_folder, output_csv):
    embeddings = []
    for file_name in os.listdir(person_folder):
        file_path = os.path.join(person_folder, file_name)
        if not file_name.lower().endswith(('.jpg', '.jpeg', '.png')):
            continue
        else:
            with torch.no_grad():
                embedding = get_osnet_1x_embedding(file_path)
                embedding = embedding.squeeze(0).cpu()
                embeddings.append(embedding)

    if len(embeddings) == 0:
        print('No valid images found')
        return
    else:
        embeddings_tensor = torch.stack(embeddings)
        avg_embedding = torch.mean(embeddings_tensor, dim = 0)
        avg_embedding_np = avg_embedding.numpy()
        df = pd.DataFrame([avg_embedding_np])
        df.to_csv(output_csv, index = False)
        print(f"Average embedding saved to {output_csv}")
```

```python
person_folder = 'osama_bin_laden'
output_csv = 'osama_bin_laden.csv'
get_avg_embedding(person_folder, output_csv)
```
